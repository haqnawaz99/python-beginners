## Section 1 — Mathematical Operators

Python has 7 mathematical operators:

| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Addition | `30 + 40` | `70` |
| `-` | Subtraction | `100 - 35` | `65` |
| `*` | Multiplication | `5 * 17` | `85` |
| `/` | Division | `10 / 3` | `3.333...` (always float) |
| `//` | Floor Division | `10 // 3` | `3` (whole part only) |
| `%` | Modulus (remainder) | `10 % 3` | `1` |
| `**` | Exponent (power) | `2 ** 3` | `8` |

````python
# Addition (+)
tafsir_books = 40
hadith_books = 30
total_books = tafsir_books + hadith_books
print("Islamic books in library:", total_books)

fajr_students = 15
dhuhr_students = 22
print("Students in two sessions:", fajr_students + dhuhr_students)
````

````python
# Subtraction (-)
total_students = 35
absent = 3
present = total_students - absent
print("Present students:", present)

budget = 5000
spent = 1350
remaining = budget - spent
print("Remaining budget:", remaining, "rupees")
````

````python
# Multiplication (*)
price_per_kg = 170
kg = 5
total_cost = price_per_kg * kg
print("Cost of sugar:", total_cost, "rupees")

students = 12
ayat_each = 5
total_ayat = students * ayat_each
print("Total ayat memorized:", total_ayat)
````

````python
# Division (/) — always gives a float
zakat_money = 10000
people = 4
share = zakat_money / people
print("Each person's share:", share, "rupees")

pages = 600
days = 30
pages_per_day = pages / days
print("Pages to read per day:", pages_per_day)
````

### `//` Floor Division vs `/` Regular Division

This is the operator students confuse most.

| | `/` | `//` |
|-|-----|------|
| `10 / 3` | `3.333...` (decimal) | — |
| `10 // 3` | — | `3` (whole number only, decimal cut off) |

> **Real example:** 35 students, 6 per group. How many **complete** groups?  
> `35 // 6 = 5` complete groups, `35 % 6 = 5` students left over.

````python
# Floor Division (//) — whole number only
qurans = 25
rows = 3
per_row = qurans // rows
print("Qurans per row:", per_row)      # 8 (not 8.33)

rupees = 1000
per_child = rupees // 7
print("Rupees per child:", per_child)
````

````python
# Modulus (%) — gives the REMAINDER
leftover_qurans = 25 % 3
print("Leftover Qurans:", leftover_qurans)    # 1

leftover_rupees = 1000 % 7
print("Leftover rupees:", leftover_rupees)

# Useful: check if a number is even or odd
students = 35
print("Is class size even?:", students % 2 == 0)   # False
````

````python
# Exponent (**) — power
print("2 ** 3 =", 2 ** 3)    # 8
print("3 ** 2 =", 3 ** 2)    # 9
print("10 ** 2 =", 10 ** 2)  # 100
````

````python
# Using operators with int(input()) and f-strings
price = int(input("Enter price per item (rupees): "))
qty = int(input("Enter quantity: "))
print(f"Total cost: {price * qty} rupees")
print(f"Each item costs {price} rupees — {qty} items = {price * qty} rupees")

print()
print("All 7 mathematical operators covered. Well done!")
````

## Section 2 — BODMAS (Order of Operations)

When a calculation has multiple operators, Python follows **BODMAS** — exactly like your maths class:

| Letter | Meaning | Operators |
|--------|---------|----------|
| **B** | Brackets | `( )` |
| **O** | Orders (powers) | `**` |
| **D** | Division | `/` `//` `%` |
| **M** | Multiplication | `*` |
| **A** | Addition | `+` |
| **S** | Subtraction | `-` |

> **Rule:** Division and Multiplication have equal priority — evaluated left to right.  
> **Tip:** When in doubt — use brackets! They make your intention clear.

````python
# Without brackets — Python decides the order
print("5 + 3 * 2  =", 5 + 3 * 2)      # 11 not 16 — * before +
print("10 - 4 * 2 =", 10 - 4 * 2)     # 2  not 12
print("6 + 8 / 4  =", 6 + 8 / 4)      # 8.0 not 3.5
````

````python
# With brackets — YOU control the order
print("(5 + 3) * 2  =", (5 + 3) * 2)      # 16
print("(10 - 4) * 2 =", (10 - 4) * 2)     # 12
print("(6 + 8) / 4  =", (6 + 8) / 4)      # 3.5
````

````python
# Powers before multiplication
print("2 * 3 ** 2  =", 2 * 3 ** 2)    # 18 (3**2=9 first, then 2*9)
print("5 + 2 ** 3  =", 5 + 2 ** 3)    # 13 (2**3=8 first, then 5+8)
print("(2 + 1) ** 3 =", (2 + 1) ** 3) # 27 (bracket first: 3**3)
````

````python
# Real Islamic / Pakistani examples
# Tasbeeh: 33 per prayer x 5 prayers, done by 3 people
total_tasbeeh = 33 * 5 * 3
print(f"Total tasbeeh (3 people, 5 prayers): {total_tasbeeh}")

# 30 paras x 20 pages each, minus 10 pages already done
remaining_pages = (30 * 20) - 10
print(f"Remaining Quran pages to read: {remaining_pages}")

# Bill split: 3 items at 250 each, split among 5 friends
bill = (3 * 250) / 5
print(f"Each friend pays: {bill} rupees")
````

````python
# Prediction challenge — predict BEFORE running!
print("10 + 5 * 3       =", 10 + 5 * 3)
print("(10 + 5) * 3     =", (10 + 5) * 3)
print("2 ** 3 + 4 * 2   =", 2 ** 3 + 4 * 2)
print("(2 ** 3 + 4) * 2 =", (2 ** 3 + 4) * 2)

print()
print("When in doubt — use brackets! They make your intention clear.")
````

## Section 3 — Comparison Operators

Comparison operators **compare two values**. The result is always `True` or `False`.

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 6` | `True` |
| `>` | Greater than | `8 > 3` | `True` |
| `<` | Less than | `3 < 8` | `True` |
| `>=` | Greater than or equal | `5 >= 5` | `True` |
| `<=` | Less than or equal | `3 <= 5` | `True` |

> ⚠️ `=` gives a value. `==` asks a question. They are completely different!

````python
# == Equal to
print("'Fajr' == 'Fajr'     ->", "Fajr" == "Fajr")        # True
print("'Maghrib' == 'Isha'  ->", "Maghrib" == "Isha")      # False
print("5 == 5               ->", 5 == 5)                    # True
print("5 == 6               ->", 5 == 6)                    # False
````

````python
# != Not equal to
print("'Ramadan' != 'Shawwal' ->", "Ramadan" != "Shawwal") # True
print("'Friday' != 'Friday'   ->", "Friday" != "Friday")   # False
print("10 != 5                ->", 10 != 5)                 # True
````

````python
# > and < Greater than and Less than
print("4 > 2  ->", 4 > 2)       # True
print("8 > 10 ->", 8 > 10)      # False
print("5 < 10  ->", 5 < 10)     # True
print("65 < 50 ->", 65 < 50)    # False
````

````python
# >= and <= Greater/less than or equal to
score = 70
passing_score = 70
print("Score >= passing score:", score >= passing_score)    # True

quran_level = 3
print("Level <= 5:", quran_level <= 5)     # True
print("12 <= 6:", 12 <= 6)                 # False
````

````python
# Comparison with int(input())
student_marks = int(input("Enter student marks (out of 100): "))
passing_marks = 50

print(f"Marks entered      : {student_marks}")
print(f"Did student pass?  : {student_marks >= passing_marks}")
print(f"Did student fail?  : {student_marks < passing_marks}")
print(f"Full marks?        : {student_marks == 100}")

print()
print("Comparison operators always give True or False. Great work!")
````

## Section 4 — Logical Operators

Logical operators **combine** two or more conditions. The result is always `True` or `False`.

| Operator | Meaning | Rule |
|----------|---------|------|
| `and` | Both must be True | `True and False` → `False` |
| `or` | At least one must be True | `True or False` → `True` |
| `not` | Reverses True/False | `not True` → `False` |

### `and` truth table
```
True  and True  → True
True  and False → False
False and True  → False
False and False → False
```

### `or` truth table
```
True  or True  → True
True  or False → True
False or True  → True
False or False → False
```

````python
# and — BOTH conditions must be True
has_attended = True
has_done_homework = True
print("Attended AND homework done:", has_attended and has_done_homework)  # True

has_attended = True
has_done_homework = False
print("Attended AND homework done:", has_attended and has_done_homework)  # False

# Gets certificate only if attended AND passed
attended = True
passed = True
gets_certificate = attended and passed
print("Gets certificate:", gets_certificate)
````

````python
# or — at least ONE condition must be True
is_reading_quran = True
is_listening_lecture = False
print("Reading OR listening:", is_reading_quran or is_listening_lecture)  # True

is_reading_quran = False
is_listening_lecture = False
print("Reading OR listening:", is_reading_quran or is_listening_lecture)  # False

# Gets extra time if sick OR has valid excuse
is_sick = False
has_valid_excuse = True
gets_extra_time = is_sick or has_valid_excuse
print("Gets extra time:", gets_extra_time)
````

````python
# not — reverses True/False
is_fasting = False
print("not is_fasting:", not is_fasting)    # True

is_weekend = True
print("not is_weekend:", not is_weekend)    # False

is_absent = False
print("Is present:", not is_absent)         # True (not absent = present)
````

````python
# Combining operators
age = 15
has_basic_knowledge = True
is_weekend = False

eligible = age >= 12 and has_basic_knowledge and not is_weekend
print("Eligible for advanced class:", eligible)

# Hifz admission: age 8-16 AND at least 5 surahs memorized
student_age = 12
surahs_memorized = 8
can_join = (student_age >= 8 and student_age <= 16) and surahs_memorized >= 5
print("Can join Hifz program:", can_join)
````

````python
# Logical operators with int(input())
age = int(input("Student age: "))
marks = int(input("Student marks (out of 100): "))

passed = marks >= 50
correct_age = age >= 10 and age <= 16
promoted = passed and correct_age

print(f"\nAge in range (10-16)  : {correct_age}")
print(f"Passed exam (50+)     : {passed}")
print(f"Eligible for promotion: {promoted}")

print()
print("Logical operators let you make complex decisions. Excellent!")
````

## Section 5 — Module Summary

### What you have learned

| Category | Operators | Returns |
|----------|-----------|---------|
| Math | `+` `-` `*` `/` `//` `%` `**` | Number |
| Comparison | `==` `!=` `>` `<` `>=` `<=` | `True` or `False` |
| Logical | `and` `or` `not` | `True` or `False` |

### Key Rules
1. `/` always gives a float — `10 / 2` = `5.0` not `5`
2. `//` gives the whole part only — `10 // 3` = `3`
3. `%` gives the remainder — `10 % 3` = `1`
4. `=` assigns, `==` compares — never mix them up
5. Comparison operators always return `True` or `False`
6. `and` needs both True; `or` needs at least one True; `not` reverses
7. Use brackets when BODMAS order is not what you want
