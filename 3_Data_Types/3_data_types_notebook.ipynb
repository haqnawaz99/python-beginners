## Section 1 — What is a Data Type?

Every value in Python has a **type**. The type tells Python what kind of data it is and what can be done with it.

There are 4 main types:

| Type | Full name | Examples | Used for |
|------|-----------|---------|----------|
| `int` | Integer | `5`, `114`, `1947` | Counting, ages, quantities |
| `float` | Float | `3.14`, `87.5`, `2.5` | Prices, marks, measurements |
| `str` | String | `"Ahmed"`, `"Lahore"` | Names, messages, text |
| `bool` | Boolean | `True`, `False` | Yes/no, on/off answers |

You can always ask Python: **"what type is this?"** using `type()`.

````python
# type() — check the type of any value
print(type(42))           # <class 'int'>
print(type(3.14))         # <class 'float'>
print(type("Ahmed"))      # <class 'str'>
print(type(True))         # <class 'bool'>
print(type(None))         # <class 'NoneType'>
````

### Type 1 — `int` (Integer)

Whole numbers — **no decimal point**.  
Used for counting, ages, quantities.

````python
# int examples
students = 35
prayers_per_day = 5
surahs_in_quran = 114
pakistan_year = 1947

print("Students:", students)
print("Daily prayers:", prayers_per_day)
print("Surahs in Quran:", surahs_in_quran)
print("Type:", type(students))           # <class 'int'>
````

### Type 2 — `float` (Decimal Number)

Numbers **with a decimal point**.  
Used for prices, marks, measurements.

````python
# float examples
price_per_kg = 135.50
temperature = 38.5
percentage = 87.5
zakat_rate = 2.5

print("Price per kg:", price_per_kg)
print("Temperature:", temperature, "C")
print("Marks percentage:", percentage)
print("Type:", type(price_per_kg))       # <class 'float'>
````

### Type 3 — `str` (String)

Text inside quotes — anything in `" "` or `' '`.  
You already know this from Module 1.

````python
# str examples
student_name = "Fatima"
city = "Lahore"
greeting = "Assalamu Alaikum"
surah = "Al-Fatiha"

print("Name:", student_name)
print("City:", city)
print("Type:", type(student_name))       # <class 'str'>
````

### Type 4 — `bool` (Boolean)

Only two possible values: **`True`** or **`False`**  
⚠️ Capital T and F — Python is case-sensitive!

````python
# bool examples
is_student = True
has_memorized_quran = False
is_fasting = True

print("Is student:", is_student)
print("Has memorized Quran:", has_memorized_quran)
print("Is fasting:", is_fasting)
print("Type:", type(is_student))         # <class 'bool'>
````

### ⚠️ `"5"` is NOT the same as `5`

This is one of the most important things to understand:

| Value | Type | What `+ ` does |
|-------|------|----------------|
| `"5"` | `str` | Joins text: `"5" + "5"` → `"55"` |
| `5` | `int` | Does maths: `5 + 5` → `10` |

````python
# "5" vs 5 — very different!
a = "5"     # string
b = 5       # integer

print(type(a))    # str
print(type(b))    # int
print(a + a)      # 55  (joins text)
print(b + b)      # 10  (adds numbers)
````

````python
# Underscores make large numbers readable
population = 220_000_000
print("Pakistan population:", population)

print()
print("You now know all 4 main data types. Ma sha Allah!")
````

## Section 2 — Type Conversion

Python does **not** mix types automatically. You must convert them yourself.

| Function | Converts to | Example |
|----------|------------|----------|
| `int(x)` | Integer | `int("25")` → `25` |
| `float(x)` | Decimal | `float("3.14")` → `3.14` |
| `str(x)` | String | `str(114)` → `"114"` |

### `str()` — number to string
Use when you want to join a number with text using `+`.

````python
# str() — convert number to string for joining with +
age = 15
print("I am " + str(age) + " years old.")

ajza = 18
print("Hafiz Ahmed memorized " + str(ajza) + " ajza.")

year = 1947
print("Pakistan was created in " + str(year) + ".")
````

### `int()` — string to integer

`input()` **always** returns a string — even if the user types `25`.  
Wrap it with `int()` to turn it into a real number for maths.

````python
# int() — see the difference
num_str = "25"
print("String + 5 =", num_str + "5")     # 255  (text joins!)
print("Integer + 5 =", int(num_str) + 5)  # 30   (maths works!)
````

### 🔑 The Most Important Pattern: `int(input())`

```python
num = int(input("Enter a number: "))
```

This does two things at once:
1. `input()` asks the user and gets a **string**
2. `int()` converts that string to a **number**

You will use this pattern in **every future module**.

````python
# int(input()) — the critical pattern
num = int(input("Enter a number: "))
print("Your number doubled:", num * 2)
print("Your number squared:", num ** 2)
````

````python
# float(input()) — for decimal numbers
marks = float(input("Enter your marks (e.g. 87.5): "))
print("Your marks:", marks)
print("Out of 100, you scored:", marks, "%")
````

````python
# Practical — shop bill calculator
price = int(input("Enter price per item (rupees): "))
quantity = int(input("How many items? "))
total = price * quantity
print("Total bill:", total, "rupees")
````

````python
# Practical — class attendance
students = int(input("Total students in class: "))
absent = int(input("How many absent today? "))
present = students - absent
print("Present:", present, "out of", students)
````

````python
# float to int — decimal part is CUT (not rounded)
marks = 89.9
print("As float:", marks)
print("As int:", int(marks))    # 89 — NOT 90, decimal is removed

print()
print("Type conversion is one of the most useful skills. Well done!")
````

## Section 3 — Conversion Errors

Type conversion can go wrong. Knowing the errors helps you fix them fast.

| Error | Cause | Fix |
|-------|-------|-----|
| `ValueError` | `int("abc")` — text is not a number | Only convert numeric strings |
| `ValueError` | `int("12.5")` — decimal string → int directly | Use `int(float("12.5"))` |
| `TypeError` | `"Age: " + 20` — string + number | Use `str(20)` or a comma |
| `TypeError` | `input() + 5` — forgot `int()` | Use `int(input(...))` |

````python
# CORRECT conversions
number = int("123")       # works — "123" looks like a whole number
print("Correct:", number)

# WRONG — uncomment one at a time to see each error:
# number = int("abc")       # ValueError — letters cannot be a number
# number = int("12 years")  # ValueError — has letters
# number = int("12.5")      # ValueError — use float() first!
````

````python
# TypeError — string + number
age = 20

# WRONG (uncomment to see the error):
# print("You are " + age + " years old.")  # TypeError

# Fix 1 — use str():
print("You are " + str(age) + " years old.")

# Fix 2 — use comma in print():
print("You are", age, "years old.")
````

````python
# Converting a decimal string to int — two steps
# WRONG (uncomment to see the error):
# x = int("45.6")           # ValueError

# CORRECT — two steps:
float_val = float("45.6")
int_val = int(float_val)
print("Float string to int:", int_val)   # 45
````

````python
# The most common beginner mistake — forgetting int() with input()
# WRONG (uncomment to see the error):
# num = input("Enter a number: ")
# print(num + 5)             # TypeError — string + int

# CORRECT:
num = int(input("Enter a number: "))
print("Your number + 5 =", num + 5)
````

````python
# Conversion rules summary
print(int("25"))           # 25    works
print(float("25.5"))       # 25.5  works
print(str(100))            # 100   works
print(int(float("25.5")))  # 25    works (two steps)

print()
print("Remember: when in doubt, check the type with type()")
x = input("Type anything: ")
print("You typed:", x)
print("Its type is:", type(x))    # always str from input()
````

## Section 4 — f-strings

An **f-string** is a cleaner way to mix variables and text together.

- Start with `f` before the opening quote
- Put variable names (or calculations) inside `{ }` curly brackets
- Python replaces `{variable}` with its actual value automatically

### Old way vs f-string

```python
name = "Ahmed"
age = 15

# Old way — messy, needs str()
print("My name is " + name + " and I am " + str(age) + " years old.")

# f-string — clean and simple
print(f"My name is {name} and I am {age} years old.")
```

> ⚠️ Don't forget the `f` before the quote — without it, curly braces print literally.

````python
# Old way vs f-string — see the difference
name = "Ahmed"
age = 15

# Old way:
print("My name is " + name + " and I am " + str(age) + " years old.")

# f-string:
print(f"My name is {name} and I am {age} years old.")
````

````python
# f-string — Islamic context
student = "Hafiz Usman"
ajza = 18
print(f"{student} has memorized {ajza} ajza of the Quran.")

surah = "Al-Fatiha"
verses = 7
print(f"Surah {surah} has {verses} verses.")

prayers = 5
print(f"A Muslim prays {prayers} times every day.")
````

````python
# f-string — Pakistani context
city = "Lahore"
population = 13_000_000
print(f"{city} has a population of about {population}.")

item = "rice"
price = 180
kg = 5
print(f"{kg} kg of {item} costs {price * kg} rupees.")

team = "Pakistan"
score = 320
print(f"{team} scored {score} runs in the match.")
````

````python
# f-strings can do calculations inside {}
marks = 85
total = 100
print(f"Marks: {marks} out of {total}")
print(f"Percentage: {marks / total * 100}%")

tasbeeh = 33
print(f"Total tasbeeh in 5 prayers: {tasbeeh * 5}")
````

````python
# f-strings with input()
your_name = input("Enter your name: ")
your_city = input("Enter your city: ")
print(f"Assalamu Alaikum {your_name}! Welcome from {your_city}.")
````

````python
# f-strings with int(input())
school = input("Your school/madrasa name: ")
teacher = input("Your teacher's name: ")
print(f"You study at {school} under {teacher}. May Allah bless you both!")

print()
print("f-strings make your code much easier to read. Excellent!")
````

## Section 5 — Module Summary

### What you have learned

| Concept | Syntax | Example |
|---------|--------|---------|
| Check type | `type(x)` | `type(42)` → `<class 'int'>` |
| String → int | `int(x)` | `int("25")` → `25` |
| String → float | `float(x)` | `float("3.5")` → `3.5` |
| Number → string | `str(x)` | `str(114)` → `"114"` |
| Get number from user | `int(input(...))` | `age = int(input("Age: "))` |
| f-string | `f"text {var}"` | `f"Hello {name}"` |

### Key Rules
1. `"5"` (string) and `5` (int) are different — `+` joins text, not adds numbers
2. `input()` **always** returns a string — use `int()` or `float()` for maths
3. `int()` only works on strings that look like whole numbers — `int("abc")` crashes
4. f-strings need `f` before the quote and `{}` around variables
5. f-strings can do calculations inside `{}` — `f"{price * qty} rupees"`
