## Section 1 — What is a Variable?

A **variable** is a named box that stores a value.

Think of it like a student's bag:
- The **name tag** on the bag is the variable name
- The **books inside** the bag are the value
- You find the bag by reading its name tag

### How to create a variable

```python
variable_name = value
```

The `=` sign means **"store this value in this name"** — it is not the same as equals in maths.

```python
greeting = "Assalamu Alaikum"
student_name = "Ahmed"
city = "Lahore"
```

### How to use a variable

Just write the variable name — **no quotes needed**:

```python
print(greeting)        # prints: Assalamu Alaikum
print(student_name)    # prints: Ahmed
```

````python
# Creating variables
greeting = "Assalamu Alaikum"
student_name = "Ahmed"
city = "Lahore"
school = "Jamia Islamia"

# Using variables in print()
print(greeting)
print(student_name)
print(city)
print(school)
````

````python
# Variables work with + just like strings
print("Hello, " + student_name)
print(student_name + " is from " + city)
print("Welcome to " + school + "!")
````

### Changing a variable

You can update a variable at any time. The **new value replaces the old one** — the box now holds something new.

````python
# Changing a variable
subject = "Quran"
print("Today's subject: " + subject)

subject = "Hadith"
print("Now the subject is: " + subject)
````

````python
# Multiple variables together
first_name = "Fatima"
last_name = "Khan"
age = "15"
city = "Karachi"

print("Name : " + first_name + " " + last_name)
print("Age  : " + age)
print("City : " + city)

print()
print("You now understand variables. Excellent work!")
````

## Section 2 — Storing `input()` in a Variable

In Module 1 we used `input()` directly inside `print()`. That works — but only once.

```python
# Module 1 style — can only use the answer once:
print("Hello, " + input("Your name? "))
```

If you store the answer in a variable, you can **use it as many times as you want**:

```python
# Module 2 style — store and reuse:
name = input("Your name? ")
print("Assalamu Alaikum, " + name + "!")
print("May Allah bless you, " + name + ".")
print(name + ", welcome to today's Python lesson.")
```

The user types their name **once** — you can use it in ten places if you want.

````python
# Without variable — can only use once
print("Hello, " + input("Your name? "))
print()
````

````python
# With variable — reuse the answer multiple times
name = input("Your name? ")
print("Assalamu Alaikum, " + name + "!")
print("May Allah bless you, " + name + ".")
print(name + ", welcome to today's Python lesson.")
````

````python
# More reuse examples
favorite_book = input("What is your favorite Islamic book? ")
print("Ma sha Allah! " + favorite_book + " is a wonderful book.")
print("May Allah help you understand " + favorite_book + " deeply.")
````

````python
surah = input("What is your favorite Surah? ")
print("Surah " + surah + " is indeed beautiful.")
print("May Allah give you tawfiq to memorize Surah " + surah + ".")
````

````python
# Building a student profile
print("Let us build your profile:")
print()

student_name = input("Your name: ")
student_city = input("Your city: ")
student_class = input("Your class/level: ")

print()
print("--- Student Profile ---")
print("Name  : " + student_name)
print("City  : " + student_city)
print("Class : " + student_class)
print("Assalamu Alaikum, " + student_name + " from " + student_city + "!")

print()
print("Well done! Storing input in variables is very useful.")
````

## Section 3 — `len()` with Variables

In Module 1 we used `len()` directly on strings like `len("Ahmed")`.  
Now that we have variables, we can use `len()` on a variable too:

```python
name = "Muhammad"
print(len(name))       # 8
```

### Two ways to print `len()` in a sentence

| Method | Code | Notes |
|--------|------|-------|
| Comma | `print("Length:", len(name))` | Easier — Python adds space automatically |
| `str()` + `+` | `print("Length: " + str(len(name)))` | More control over formatting |

````python
# len() with variables
name = "Muhammad"
print(len(name))               # 8

city = "Islamabad"
print(len(city))               # 9

surah = "Al-Fatiha"
print(len(surah))              # 9
````

````python
# Printing len() in a sentence — two ways
name = "Abdullah"

# Using comma (easier for beginners):
print("The name", name, "has", len(name), "letters.")

# Using str() and +:
print("The name " + name + " has " + str(len(name)) + " letters.")
````

````python
# len() with input()
word = input("Type any word: ")
print("You typed:", word)
print("It has", len(word), "characters.")
````

````python
# len() on Islamic words
bismillah = "Bismillahirrahmanirrahim"
print("Bismillah has", len(bismillah), "characters (no spaces)")

kalima = "La ilaha illallah Muhammadur Rasulullah"
print("The Kalima has", len(kalima), "characters including spaces")

print()
print("Great! You can now measure the length of any string.")
````

````python
# len() with a Surah name from user
surah_name = input("Enter a Surah name: ")
print("Surah " + surah_name + " has", len(surah_name), "characters in its name.")
````

## Section 4 — Variable Naming Rules

Python has strict rules about what you can name a variable.

### ✅ Rules — you MUST follow these

| Rule | Example |
|------|---------|
| Only letters, numbers, underscores | `student_name`, `city1`, `_temp` |
| Must START with a letter or `_` (not a number) | `name` ✅ — `1name` ❌ |
| No spaces — use underscore instead | `student_name` ✅ — `student name` ❌ |
| Cannot use Python keywords | `for`, `if`, `print` are reserved |
| Case-sensitive | `name`, `Name`, `NAME` are **three different variables** |

### ❌ Wrong names that cause errors

```python
student name = "Ahmed"    # SyntaxError — space not allowed
1city = "Lahore"          # SyntaxError — starts with a number
student-name = "Ahmed"    # SyntaxError — hyphen not allowed
print = "something"       # Breaks the print function!
```

### 👍 Good naming habits

| Bad | Good | Why |
|-----|------|-----|
| `x` | `student_city` | Name tells you what it stores |
| `sn` | `school_name` | Easy to read |
| `myVar` | `my_var` | Python convention: use underscores |

````python
# CORRECT variable names
student_name = "Ahmed"
city1 = "Lahore"
_my_var = "test"
firstName = "Fatima"
SCHOOL = "Jamia Islamia"
total_students = "50"

print(student_name)
print(city1)
print(firstName)
print(SCHOOL)
````

````python
# WRONG names — uncomment one at a time to see each error

# student name = "Ahmed"       # SyntaxError — space
# 1city = "Lahore"             # SyntaxError — starts with number
# student-name = "Ahmed"       # SyntaxError — hyphen
# for = "loop"                 # SyntaxError — reserved word

print("Uncomment a line above to see what error it gives.")
````

````python
# Python is case-sensitive — these are THREE different variables
name = "Ahmed"
Name = "Ali"
NAME = "Hassan"

print(name)     # Ahmed
print(Name)     # Ali
print(NAME)     # Hassan
````

````python
# Good naming examples
favorite_surah = "Al-Fatiha"
prayer_time = "Fajr"
total_students_in_class = "35"

print("Favorite Surah : " + favorite_surah)
print("Prayer time   : " + prayer_time)
print("Total students: " + total_students_in_class)

print()
print("Remember: good variable names make your code easy to read!")
````

## Section 5 — String Methods

**Methods** are built-in actions you can perform on a string.  
You call them by putting a **dot (`.`)** after the string or variable:

```python
name = "ahmed"
print(name.upper())    # AHMED
```

> ⚠️ Methods do **not** change the original variable — they return a **new** value.

### Methods covered in this module

| Method | What it does | Example |
|--------|-------------|----------|
| `.upper()` | All CAPITALS | `"ahmed".upper()` → `"AHMED"` |
| `.lower()` | all lowercase | `"AHMED".lower()` → `"ahmed"` |
| `.title()` | First Letter Of Each Word | `"ahmed ali".title()` → `"Ahmed Ali"` |
| `.strip()` | Remove spaces from both sides | `"  Ahmed  ".strip()` → `"Ahmed"` |
| `.lstrip()` | Remove spaces from left only | `"  Ahmed".lstrip()` → `"Ahmed"` |
| `.rstrip()` | Remove spaces from right only | `"Ahmed  ".rstrip()` → `"Ahmed"` |

````python
# .upper() — converts to ALL CAPITALS
name = "ahmed ali"
print(name.upper())          # AHMED ALI

school = "jamia islamia lahore"
print(school.upper())        # JAMIA ISLAMIA LAHORE
````

````python
# .lower() — converts to all lowercase
name = "FATIMA KHAN"
print(name.lower())          # fatima khan

email = "Student@Madrasa.EDU"
print(email.lower())         # student@madrasa.edu
````

````python
# .title() — capitalizes the First Letter of Each Word
name = "maulana abdur rahman"
print(name.title())          # Maulana Abdur Rahman

city = "lahore pakistan"
print(city.title())          # Lahore Pakistan

book = "sahih al-bukhari"
print(book.title())          # Sahih Al-Bukhari
````

````python
# .title() is very useful with input() — cleans up whatever the user types
raw_name = input("Enter your name (any format): ")
clean_name = raw_name.title()
print("Formatted name: " + clean_name)
````

````python
# .strip() — removes accidental spaces from both sides
messy = "   Ahmed   "
print(messy.strip())         # Ahmed

with_tabs = "\tLahore\t"
print(with_tabs.strip())     # Lahore
````

````python
# .lstrip() and .rstrip() — one side only
left_spaces = "   Karachi"
print(left_spaces.lstrip())  # Karachi  (removed left spaces only)

right_spaces = "Peshawar   "
print(right_spaces.rstrip()) # Peshawar  (removed right spaces only)
````

````python
# Chaining methods — use more than one at a time
raw_input = "   maulana tariq jameel   "
clean = raw_input.strip().title()
print(clean)                 # Maulana Tariq Jameel
````

### Special characters: `\t` and `\n`

| Character | What it does | Example |
|-----------|-------------|----------|
| `\t` | Tab (large space) | `"Name:\tAhmed"` → `Name:    Ahmed` |
| `\n` | New line | `"Line1\nLine2"` → prints on two lines |

````python
# \t and \n for neat formatting
print("Name:\tAhmed")
print("City:\tLahore")
print()
print("Line 1\nLine 2\nLine 3")

print()
print("String methods make your data clean and consistent. Well done!")
````

### Puzzle 1 — What gets printed?

A variable is assigned three times. What is printed at the end?

````python
# Puzzle 1 — predict before running!
city = "Lahore"
city = "Karachi"
city = "Islamabad"
print(city)

# What do you think prints? Why?
````

### Puzzle 2 — Swapping Two Variables

We have two variables. We want to swap their values using a **temporary variable**.

````python
# Puzzle 2 — swap
first = "Fajr"
second = "Isha"

print("Before swap:")
print("first  =", first)
print("second =", second)

# Swap using a temporary variable
temp = first
first = second
second = temp

print("After swap:")
print("first  =", first)
print("second =", second)
````

### Puzzle 3 — Building a Full Name

````python
# Puzzle 3 — build and rebuild a full name
title = "Maulana"
first_name = "Tariq"
last_name = "Jameel"

full_name = title + " " + first_name + " " + last_name
print(full_name)

# Now change first_name and rebuild
first_name = "Abdur Rahman"
full_name = title + " " + first_name + " " + last_name
print(full_name)
````

### Puzzle 4 — Chain Swap

Three variables. Each one shifts to the next. Predict the final order.

````python
# Puzzle 4 — chain swap
a = "Quran"
b = "Hadith"
c = "Fiqh"

print("Before:", a, b, c)

temp = a
a = b
b = c
c = temp

print("After :", a, b, c)
````

### Puzzle 5 — Does `y` change when `x` changes?

````python
# Puzzle 5 — prediction challenge
x = "Pakistan"
y = x
x = "India"

print(x)
print(y)

# Did y change when x changed? Why or why not?
````
