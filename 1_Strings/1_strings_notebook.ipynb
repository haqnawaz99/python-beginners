## Section 1 — What is a String?

A **string** is any piece of text inside quotation marks.

It can contain:
- Letters: `"Ahmed"`
- Words: `"Assalamu Alaikum"`
- Sentences: `"The Quran has 114 Surahs"`
- Numbers written as text: `"786"`
- Urdu or Arabic: `"السلام علیکم"`

Python treats **everything inside quotation marks as text** — it will never try to calculate it.

### Single quotes or double quotes?

Both work. `"Ahmed"` and `'Ahmed'` are exactly the same.  
Pick one style and stay consistent.


> We will learn about numbers vs strings in **Module 3 (Data Types)**.


````python
# Run this cell — these are all strings
print("Assalamu Alaikum")
print("Pakistan")
print("123")           # Even numbers become text when inside quotes
print("!")
print("Surah Al-Fatiha has 7 verses")
````

````python
# Single quotes and double quotes both work
print("This uses double quotes")
print('This uses single quotes')
# Both lines print the same kind of output
````

````python
# Strings can be short or very long
print("Hi")
print("The Holy Quran was revealed to Prophet Muhammad (PBUH) over 23 years")
print("Lahore")
print("May Allah bless you and your family. Ameen.")

print()
print("You now understand what a string is. Ma sha Allah!")
````

## Section 2 — String Concatenation (Joining Strings)

**Concatenation** means joining two or more strings into one.  
We use the **`+`** symbol to join strings.

> Think of it like gluing words together side by side.

```python
"Assalamu" + " Alaikum"   →   "Assalamu Alaikum"
```

### ⚠️ Spaces must be added manually

The `+` operator joins **exactly** what you give it — nothing more.

| Code | Output |
|------|--------|
| `"Assalamu" + "Alaikum"` | `AssalamuAlaikum` ❌ |
| `"Assalamu " + "Alaikum"` | `Assalamu Alaikum` ✅ |
| `"Assalamu" + " Alaikum"` | `Assalamu Alaikum` ✅ |

The space belongs inside one of the strings.

````python
# Basic concatenation
print("Assalamu" + " Alaikum")
print("Maulana " + "Abdul Rahman")
print("Lahore" + ", " + "Pakistan")
````

````python
# See what happens when you forget the space
print("Assalamu" + "Alaikum")      # Wrong — no space
print("Assalamu " + "Alaikum")     # Correct — space at the end of first string
print("Assalamu" + " Alaikum")     # Also correct — space at the start of second string
````

````python
# Practical examples with Pakistani/Islamic context
print("Welcome, " + "Ustaz Ahmad!")
print("In the name of " + "Allah, the Most Gracious, the Most Merciful")
print("Today's lesson: " + "Introduction to Python")
print("I am from " + "Karachi" + ", Pakistan")
print("The Quran has " + "114" + " Surahs and " + "30" + " Paras")
````

### Concatenation with Urdu and Arabic

Python supports all languages including **Urdu** and **Arabic**.  
The same `+` operator works exactly the same way.

> Note: In the terminal, right-to-left text may look slightly different, but the logic is identical to English strings.

````python
# Urdu strings
print("السلام" + " علیکم")
print("خوش آمدید " + "میرے پیارے طالبعلموں")
print("مولانا " + "عبدالرحمن")
print("ہم " + "Python " + "سیکھ رہے ہیں")
print("محنت " + "کامیابی کی کنجی ہے")
print("علم حاصل کرنا " + "ہر مسلمان پر فرض ہے۔")
````

````python
# Arabic strings
print("بِسْمِ اللّٰہِ " + "الرَّحْمٰنِ الرَّحِیْمِ")
print("اقْرَأْ بِاسْمِ رَبِّكَ " + "الَّذِي خَلَقَ")
````

````python
# Mixing English and Urdu
print("ہم " + "Python " + "سیکھ رہے ہیں")      # Urdu + English
print("My name is " + "احمد علی")               # English + Urdu
print("Subject: " + "علوم القرآن")              # English label + Urdu value

print()
print("Excellent! Python speaks Urdu and Arabic too. Alhamdulillah!")
````

### ⚠️ Common Mistake — You cannot `+` a string and a number

This is one of the most common errors beginners make:

```python
print("There are " + 5 + " prayers")   # ❌ TypeError!
```

Python does not know how to glue the **text** `"There are "` to the **number** `5`.  
You must first convert the number to a string using **`str()`**.

**Two ways to fix this:**

| Method | Code | Notes |
|--------|------|-------|
| `str()` conversion | `"There are " + str(5) + " prayers"` | Works everywhere |
| Comma in `print()` | `print("There are", 5, "prayers")` | Easier but adds spaces automatically |

````python
# This will cause a TypeError — uncomment the next line to see the error
# print("There are " + 5 + " prayers")

# Fix 1 — use str() to convert the number to a string
print("There are " + str(5) + " prayers")

# Fix 2 — use a comma inside print() instead of +
print("There are", 5, "prayers")
````

````python
# Another example
print("Pakistan was created in " + str(1947))
print("The Quran has " + str(114) + " Surahs")
````

## Section 3 — `input()` — Getting Information from the User

So far, all the information in our programs has been **fixed** — we wrote it ourselves.  
With `input()`, we can ask the **user** to type something.

### How `input()` works

```python
name = input("Please enter your name: ")
```

1. Python **stops** the program
2. It **shows** the text inside `input("...")` as a prompt
3. It **waits** for the user to type and press Enter
4. It **stores** what was typed in the variable (`name`)

### ⚠️ Important rule — `input()` always returns a STRING

Even if the user types `25`, Python receives it as `"25"` (text, not a number).  
We will learn how to convert it to a real number in **Module 3**.

### The 3-step pattern

```python
# Step 1: Ask and store
name = input("Please enter your name: ")

# Step 2: Use the variable
print("Assalamu Alaikum, " + name)
```

````python
# Basic input — ask name and greet
name = input("Please enter your name: ")
print("Assalamu Alaikum, " + name)
````

````python
# Ask for city
city = input("Which city are you from? ")
print("Ma sha Allah, " + city + " is a wonderful city!")
````

````python
# Ask for favourite subject
subject = input("What is your favorite subject? ")
print("Great choice! " + subject + " is very important.")
````

### Using  directly inside 

You can skip the variable and put  straight inside .
This is shorter, but you cannot reuse the value later.



````python
# You can also use input() directly inside print() — without a variable
# This is shorter but less flexible
print("Wa Alaikum Assalam, " + input("Who are you replying to? "))
````

````python
# Islamic and Pakistani context examples
school = input("What is the name of your school or madrasa? ")
print("May Allah give barakah to " + school + "!")
````

````python
prophet_name = input("Name your favorite Prophet (peace be upon them): ")
print("Peace and blessings be upon Prophet " + prophet_name)
````

````python
# input() always returns a STRING — even when the user types a number
age = input("How old are you? ")
print("You are " + age + " years old. May Allah bless you!")
# Notice: we can join 'age' with strings using + because it IS already a string
````

### Asking multiple questions

You can call `input()` as many times as you need — each call asks one question.

````python
# Ask first name and last name, print them joined
first = input("Enter your first name: ")
last = input("Enter your last name: ")
print("Assalamu Alaikum, " + first + " " + last + "!")
````

````python
# A simple introduction — ask questions one by one
name = input("Your name: ")
print("My name is " + name + ".")

city = input("Your city: ")
print("I am from " + city + ".")

school = input("Your school or madrasa: ")
print("I study at " + school + ".")
````

## Section 4 — `len()` — Counting Letters in a String

**`len()`** stands for *length*. It counts how many **characters** are in a string.

A character is any single letter, number, space, or symbol.

```python
len("Ahmed")           →   5
len("Alhamdulillah")   →   13
len("Hi")              →   2
len("A B")             →   3   (the space counts too!)
```

### How to use `len()`

```python
# Option 1: print the length directly
print(len("Ahmed"))

# Option 2: store it in a variable first
name = "Ahmed"
length = len(name)
print(length)
```

> **Remember:** `len()` gives back a **number** (integer). To join it with a string using `+`, you must wrap it in `str()`: `str(len(name))`

````python
# Basic len() examples
print(len("Ahmed"))            # 5
print(len("Alhamdulillah"))    # 13
print(len("Pakistan"))         # 8
print(len("Quran"))            # 5
````

````python
# Spaces count as characters too
print(len("Ali"))              # 3
print(len("Ali Khan"))         # 8  (space counts!)
````

````python
# len() with a string directly
print("The word Bismillah has " + str(len("Bismillah")) + " letters")
print("The word Alhamdulillah has " + str(len("Alhamdulillah")) + " letters")
````

````python
# len() with input()
name = input("Enter your name: ")
print("Your name has " + str(len(name)) + " letters.")
````

````python
# Practical example — check name length
name = input("Enter your full name: ")
print("Assalamu Alaikum, " + name + "!")
print("Your full name has " + str(len(name)) + " characters (including the space).")
````

## Section 5 — Common Mistakes Summary

These are the most frequent mistakes in this module. Study each one carefully.

| # | Mistake | Wrong | Fixed |
|---|---------|-------|-------|
| 1 | Forgetting space between words | `"Ahmed"+"Khan"` → `AhmedKhan` | `"Ahmed "+"Khan"` |
| 2 | Joining string + number with `+` | `"Age: " + 15` → TypeError | `"Age: " + str(15)` |
| 3 | Missing quotes in `input()` | `input(Enter name:)` → SyntaxError | `input("Enter name:")` |
| 4 | Not storing `input()` result | `input("Name: ")` | `name = input("Name: ")` |
| 5 | Using `len()` result directly with `+` | `"Length: " + len(x)` → TypeError | `"Length: " + str(len(x))` |

````python
# Mistake 1 — forgetting space
# Wrong:
print("Ahmed" + "Khan")       # AhmedKhan
# Fixed:
print("Ahmed " + "Khan")      # Ahmed Khan
````

````python
# Mistake 2 — joining string and number
# Wrong (uncomment to see the error):
# print("Age: " + 15)

# Fixed:
print("Age: " + str(15))
````

````python
# Mistake 3 — missing quotes in input()
# Wrong (uncomment to see the error):
# name = input(Enter your name: )

# Fixed:
name = input("Enter your name: ")
print(name)
````

````python
# Mistake 5 — using len() result with +
# Wrong (uncomment to see the error):
# print("Length: " + len("Bismillah"))

# Fixed:
print("Length: " + str(len("Bismillah")))
````

5. `len()` returns a number — use `str(len(...))` when joining with text.
